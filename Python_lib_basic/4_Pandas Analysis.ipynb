```python
!pip install -U datasets
import pandas as pd
from datasets import load_dataset

dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()
```

---

## 🟩 Calculate Mean Salary (1.26.1)

```python
df['salary_year_avg'].mean()
```

---

## 🟩 Calculate Median Salary (1.26.2)

```python
df['salary_year_avg'].median()
```

```python
df.info()
```

---

## 🟨 Average Hourly Salary by Job Title (1.26.3)

```python
df.groupby('job_title')['salary_hour_avg'].mean()
```

---

## 🟨 Job with Lowest Salary (1.26.4)

```python
df = df[df['salary_year_avg'].notna()]
```

```python
lowest_paid_jobs = df.groupby('job_title', as_index=False)['salary_year_avg'].mean().sort_values('salary_year_avg', ascending=True)
display(lowest_paid_jobs.head(10))
```
