## 🟨 Highest and Lowest Salary (1.22.3)

```python
# Code Needed
salaries_list = [70000, 85000, 60000, 95000, 80000]
```

```python
#solution
import numpy as np
salaries_list = [70000, 85000, 60000, 95000, 80000]
```

---

```python
np.array(salaries_list)
```

```python
m=np.array(salaries_list).max()
print(m)
```

```python
mi= np.array(salaries_list).min()
print(mi)
```
