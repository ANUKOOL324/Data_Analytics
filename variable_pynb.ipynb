## 🏗️ Python's Nature: The Object Model

### **1. The Golden Rule**
**Variables are Labels, NOT Boxes.**
* In Python, the **Object** (the data) exists in memory first.
* The **Variable** is just a name (pointer) pointing to that object's address.
* `x = 500` means "Point the name `x` to the object `500`."

### **2. First-Class Everything**
* **Data** (integers, strings) are objects.
* **Logic** (functions) are objects.
* **Blueprints** (classes) are objects.
* **Modules** (libraries like `math`) are objects.
* *Consequence:* You can pass a function into another function just like a number.



---

## 📈 The Reference Count (Memory Management)

### **What is it?**
A "Popularity Meter" for every object in memory. It counts how many variables are currently pointing to it.

| Action | Effect on Count |
| :--- | :--- |
| `a = 500` | **+1** (Object 500 has 1 fan) |
| `b = a` | **+1** (Object 500 has 2 fans) |
| `a = "Hi"` | **-1** (Object 500 has 1 fan left) |
| `del b` | **-1** (Object 500 has 0 fans) |

### **The Garbage Collector (GC)**
* When the **Reference Count hits 0**, the object is "dead."
* Python automatically deletes it from RAM to save space.



---

## ❄️ Mutability: Can I change it?

1.  **Immutable (Unchangeable):** `int`, `str`, `tuple`.
    * If you "change" a string, Python actually creates a brand new string object and moves the label.
2.  **Mutable (Changeable):** `list`, `dict`, `set`.
    * You can change the contents without moving the label to a new memory address.

---

### **Quick Revision Commands**
* `id(obj)`: See the memory address.
* `type(obj)`: See what class the object belongs to.
* `sys.getrefcount(obj)`: See the "Popularity Meter" (Reference Count).

When you write x = 5:

1.Python creates an object in memory of the class int.

##Functions are Objects (The "Logic" Object)
In other languages, a function is just a block of code. In Python, a function is an instance of the function class.

The Theory: Since it's an object, it has an ID, a type, and you can stick it into a variable or a list.

````python
def greet(name):
    return f"Hello, {name}!"

# 1. Check its identity (it lives at a memory address)
print(id(greet))

# 2. Assign it to a new name (nickname)
say_hello = greet

# 3. Put it in a list! (Try doing THIS in C++ easily)
funcs = [greet, str.upper, len]

for f in funcs:
    print(f("Partner")) # Calls greet, then upper, then len
````

##Types/Classes are Objects (The "Blueprint" Object)
This is the one that blows people's minds. Even the word int or str is an object.

The Theory: A class is an object that describes how to create other objects. It is an instance of a "Metaclass."

````python
number = 200

# 1. Everything has a type
print(type(number)) # Output: <class 'int'>

# 2. The CLASS itself is an object you can point to!
my_type = int
new_number = my_type("500") # Uses the 'int' object to make a new integer
print(new_number + 10) # Output: 510
````

##Modules are Objects (The "File" Object)
When you import math, you aren't just getting access to math; you are creating a module object.

The Theory: Importing a library creates a giant object in memory that contains all the functions of that library as "attributes."

````python
import math

# 'math' is a variable pointing to a module object
print(type(math)) # Output: <class 'module'>
print(id(math))   # It has a memory address!

# You can even pass the whole library into a function
def show_off(library):
    print(library.pi)

show_off(math) # Passing a whole module object as an argument
````

**Immutable** (Unchangeable)
Examples: int, float, string, tuple.

**Behavior**: You cannot change the object itself. If you have x = 5 and then x = x + 1, Python doesn't change the 5 into a 6. It creates a new 6 object and moves the x label to it.

**Behavior**: You can change the object without moving the label.

# 🏗️ Revision Guide: Python’s "Meta" Object Nature

In Python, the phrase **"Everything is an Object"** applies even to the code you write yourself. When you create a class, you aren't just writing text; you are creating a living entity in memory.

---

## 1. The Two Levels of Objects
When you define a class, Python creates a **Class Object** to store that definition.

```python
class Dog:
    species = "Canine"  # Class Attribute

# 1. 'Dog' is a VARIABLE pointing to a 'type' object
print(type(Dog)) # Output: <class 'type'>

# 2. 'Dog' has its own unique memory address
print(id(Dog))
```

### 🔍 The Key Distinction
* **The Class Object (`Dog`):** This is the **Blueprint**. It lives in memory once you define it. Its type is `type`.
* **The Instance Object (`my_dog`):** This is the **House**. When you call `my_dog = Dog()`, you create a new object based on that blueprint. Its type is `Dog`.



---

## 2. Classes are "First-Class"
Because the class itself is an object, you can treat it like any other variable:
* **Pass it:** You can pass the class `Dog` into a function as an argument.
* **Store it:** You can put the class `Dog` into a list or dictionary.
* **Rename it:** You can do `MyAnimal = Dog`. Now `MyAnimal()` creates a Dog.

---

## 3. Anatomy of a Class Object
A class object consists of three main parts that define behavior:

### **A. Attributes (Data)**
* **Class Attributes:** Defined directly in the class. Shared by **ALL** instances (e.g., `species = "Canine"`).
* **Instance Attributes:** Defined inside `__init__`. Unique to **each** specific object (e.g., `self.name = "Rex"`).

### **B. Methods (Behavior)**
Methods are just **function objects** that live inside the class object.
* **`self`**: When you call `my_dog.bark()`, Python automatically passes the instance as the first argument (`self`). This is how the "Machine" knows which specific "Data" to work on.

### **C. The Constructor (`__init__`)**
A special method called automatically during instantiation to initialize the object's unique data.

---

## 4. The "Metaclass" Secret
If `my_dog` is an object of class `Dog`, and `Dog` is an object of class `type`... what is `type`?

In Python, **`type`** is a **Metaclass**. It is the "Object that creates Objects that create Objects."
* Everything eventually leads back to `type` and `object`.
* **The Benefit:** Since the class is an object, you can add a new method to a class *after* it’s already defined, and all existing instances will instantly "learn" that new behavior.



---

## 5. Revision Summary Table

| Entity | What it is | Type |
| :--- | :--- | :--- |
| `my_dog = Dog()` | **Instance Object** | `Dog` |
| `class Dog:` | **Class Object** | `type` |
| `type` | **Metaclass** | `type` (It points to itself!) |

> **💡 Revision Analogy:** If a **List** is a container for numbers, a **Class** is a container for logic and data structures. Both are objects you can move, rename, and inspect.

````python
class Student:
    # 🔸 Class Variable (shared by all objects)
    school = "ABC School"

    # 🔸 Constructor (runs when object is created)
    def __init__(self, name, marks):
        # 🔸 Instance Variables (unique to each object)
        self.name = name
        self.marks = marks

    # 🔸 Instance Method
    def display(self):
        print("Name:", self.name)
        print("Marks:", self.marks)
        print("School:", Student.school)

    # 🔸 Method to update marks
    def update_marks(self, new_marks):
        self.marks = new_marks

    # 🔸 Class Method
    @classmethod
    def change_school(cls, new_school):
        cls.school = new_school

    # 🔸 Static Method
    @staticmethod
    def is_pass(marks):
        return marks >= 40

    # 🔸 Special Method
    def __str__(self):
        return f"{self.name} scored {self.marks}"


# ===============================
# 🔹 OBJECT CREATION (IMPORTANT)
# ===============================

s1 = Student("Rahul", 85)
s2 = Student("Anjali", 90)


# ===============================
# 🔹 METHOD CALLING
# ===============================

# ✅ Instance Method (using object)
s1.display()
s2.display()

# ✅ Modify instance data
s1.update_marks(95)
s1.display()

# ✅ Access instance variables
print(s1.name)
print(s1.marks)

# ✅ Access class variable
print(Student.school)
print(s1.school)


# ===============================
# 🔹 CLASS METHOD CALLING
# ===============================

Student.change_school("XYZ School")

# check change
s1.display()
s2.display()


# ===============================
# 🔹 STATIC METHOD CALLING
# ===============================

print(Student.is_pass(85))   # True
print(Student.is_pass(30))   # False


# ===============================
# 🔹 SPECIAL METHOD CALL
# ===============================

print(s1)   # calls __str__


# ===============================
# 🔹 EXTRA (IMPORTANT CONCEPTS)
# ===============================

# Dynamically adding attribute
s1.age = 20
print(s1.age)

# Delete attribute
del s1.age

# Delete object
# del s1
````

## Lambda Functions (Anonymous Functions)

A **lambda function** in Python is a small, anonymous function defined with the `lambda` keyword. It can take any number of arguments but can only have one expression.

**Syntax:**

`lambda arguments: expression`

**Key Characteristics:**

1.  **Anonymous:** They don't have a name like regular functions defined with `def`.
2.  **Single Expression:** The body of a lambda function is a single expression, the result of which is implicitly returned.
3.  **Concise:** They are often used for short, simple operations where a full `def` function would be overkill.

**When to use them:**

Lambda functions are typically used as arguments to higher-order functions (functions that take other functions as arguments), such as `filter()`, `map()`, and `sorted()`.

**Example from your notebook:**

```python
nums = [1, 2, 3, 4, 5, 6, 7]

even = list(filter(lambda x: x % 2 == 0, nums))
print(even)
```

In this example:

*   `lambda x: x % 2 == 0` is the lambda function.
*   `x` is the argument.
*   `x % 2 == 0` is the single expression. It evaluates to `True` if `x` is even, and `False` otherwise.
*   `filter()` applies this lambda function to each item in `nums` and returns an iterator of items for which the lambda returned `True` (i.e., the even numbers).

Let's add another simple example to illustrate the basic syntax:

````python
# A simple lambda function to add two numbers
add_two_numbers = lambda a, b: a + b

result = add_two_numbers(5, 3)
print(f"Result of adding 5 and 3: {result}")

# Another example: using lambda with map to square all numbers in a list
numbers = [1, 2, 3, 4, 5]
squared_numbers = list(map(lambda x: x**2, numbers))
print(f"Squared numbers: {squared_numbers}")
````

## The `sorted()` Function Explained

The `sorted()` function in Python is a built-in function that returns a **new sorted list** from the items of any iterable (like a list, tuple, string, etc.). It does not modify the original iterable.

**Basic Syntax:**

```python
sorted(iterable, *, key=None, reverse=False)
```

**Parameters:**

*   `iterable`: (Required) The sequence (list, tuple, string) or collection (set, dictionary) to sort.
*   `key`: (Optional) A function to be called on each list element prior to making comparisons. This is where `lambda` functions are frequently used for custom sorting logic.
*   `reverse`: (Optional) A Boolean value. If `True`, the list is sorted in descending order. By default, it's `False` (ascending order).

### How the `key` argument works (especially with `lambda`)

The `key` argument is incredibly powerful because it allows you to define a custom sorting criterion. Instead of sorting the elements directly, `sorted()` will sort them based on the **result of applying the `key` function to each element**.

**Your example:**

```python
nums = [23, 45, 12, 37, 89]
sorted_nums = sorted(nums, key=lambda x: x % 10)
```

Here's a breakdown:

1.  **`nums = [23, 45, 12, 37, 89]`**: This is your list of numbers.
2.  **`key=lambda x: x % 10`**: This lambda function is applied to each number `x` in `nums`.
    *   `x % 10` calculates the remainder when `x` is divided by 10. Essentially, it gives you the **last digit** of the number.
    *   For example:
        *   `23 % 10` is `3`
        *   `45 % 10` is `5`
        *   `12 % 10` is `2`
        *   `37 % 10` is `7`
        *   `89 % 10` is `9`
3.  **Sorting Process**: The `sorted()` function uses these calculated last digits (`3, 5, 2, 7, 9`) to determine the order of the original numbers. It sorts the numbers based on these derived values in ascending order by default.
    *   The smallest last digit is `2` (from `12`).
    *   Then `3` (from `23`).
    *   Then `5` (from `45`).
    *   Then `7` (from `37`).
    *   Then `9` (from `89`).

So, the output `[12, 23, 45, 37, 89]` means that the original numbers were ordered according to their last digit, from smallest to largest.

Let's run your example to see this in action.

````python
nums = [23, 45, 12, 37, 89]
sorted_nums = sorted(nums, key=lambda x: x % 10)
print(f"Original numbers: {nums}")
print(f"Sorted by last digit: {sorted_nums}")

# Let's also see it sorted in reverse by the last digit
reversed_sorted_nums = sorted(nums, key=lambda x: x % 10, reverse=True)
print(f"Sorted by last digit (descending): {reversed_sorted_nums}")
````

## Positional and Keyword Arguments in Python

When you call a function in Python, you pass values (arguments) to its parameters. These arguments can be passed in two main ways:

### 1. Positional Arguments

*   **Definition:** Arguments whose values are assigned based on their position or order in the function call.
*   **Behavior:** The first argument in the call maps to the first parameter in the function definition, the second to the second, and so on.
*   **Characteristic:** You must provide them in the correct order.

**Example:**

```python
def describe_person(name, age, city):
    print(f"Name: {name}, Age: {age}, City: {city}")

describe_person("Alice", 30, "New York")
# "Alice" is positional for 'name'
# 30 is positional for 'age'
# "New York" is positional for 'city'
```

### 2. Keyword Arguments

*   **Definition:** Arguments whose values are assigned based on the keyword (parameter name) they are explicitly tied to in the function call.
*   **Behavior:** You explicitly name which parameter each value corresponds to.
*   **Characteristic:** The order of keyword arguments does not matter, as long as they appear after any positional arguments.

**Example:**

```python
def describe_person(name, age, city):
    print(f"Name: {name}, Age: {age}, City: {city}")

describe_person(age=25, city="London", name="Bob")
# Here, 'age', 'city', and 'name' are keywords. Order doesn't change meaning.
```

### Mixing Positional and Keyword Arguments

*   **Rule:** Any positional arguments must come *before* any keyword arguments in the function call.

**Valid Example:**

```python
describe_person("Charlie", age=35, city="Paris")
# "Charlie" is positional, 'age' and 'city' are keywords.
```

**Invalid Example (would cause an error):**

```python
describe_person(age=35, "Charlie", city="Paris") # SyntaxError: positional argument follows keyword argument
```

Let's demonstrate these concepts with a runnable code example.

````python
def car_details(make, model, year, color='Unknown'):
    """
    Displays details about a car.
    'make', 'model', 'year' are required positional arguments.
    'color' is an optional keyword argument with a default value.
    """
    print(f"--- Car Details ---")
    print(f"Make: {make}")
    print(f"Model: {model}")
    print(f"Year: {year}")
    print(f"Color: {color}")
    print("-------------------")

# --- 1. Pure Positional Arguments ---
print("Calling with pure positional arguments:")
car_details("Toyota", "Camry", 2020)

# --- 2. Pure Keyword Arguments ---
print("\nCalling with pure keyword arguments (order doesn't matter):")
car_details(year=2022, make="Honda", model="Civic", color="Blue")
car_details(color="Red", model="Accord", make="Honda", year=2021) # Different order

# --- 3. Mixing Positional and Keyword Arguments (Positional first) ---
print("\nCalling with mixed arguments (positional first):")
car_details("Ford", "Focus", 2018, color="Silver")
car_details("Nissan", "Altima", year=2019, color="Black") # 'year' is keyword

# --- 4. Using default values ---
print("\nCalling, letting 'color' use its default value:")
car_details("Tesla", "Model 3", 2023)
````

## Clarifying 'Normal Arguments' and Argument Order

When you refer to 'normal arguments,' you're generally talking about the parameters defined in a function without any special prefixes like `*` or `**` (which are used for `*args` and `**kwargs`, covered in more advanced topics).

Within these 'normal arguments,' the two primary ways to pass values are indeed **positional arguments** and **keyword arguments**, which we just discussed. The key takeaway about order depends on which type you're using:

### Does Order Matter?

1.  **For Positional Arguments: YES, order matters absolutely.**
    *   When you call `my_function(value1, value2)`, `value1` is matched to the first parameter and `value2` to the second, purely by their position.
    *   If you swap them, the meaning of the function call changes (or it might even lead to an error if the types don't match the expected parameter types).

    ```python
    def introduce(name, age):
        print(f"Hello, my name is {name} and I am {age} years old.")

    introduce("Alice", 30) # Correct: name=Alice, age=30
    introduce(30, "Alice") # Incorrect: name=30, age=Alice (likely not desired)
    ```

2.  **For Keyword Arguments: NO, order does NOT matter.**
    *   When you use `my_function(param_name=value, another_param=another_value)`, Python knows exactly which parameter gets which value because you explicitly name them.

    ```python
    def introduce(name, age):
        print(f"Hello, my name is {name} and I am {age} years old.")

    introduce(age=30, name="Bob") # Works perfectly, order doesn't matter
    introduce(name="Charlie", age=25) # Also works perfectly
    ```

3.  **When Mixing Positional and Keyword Arguments: YES, the order of *types* matters.**
    *   As a strict rule in Python, **all positional arguments must come before any keyword arguments** in a function call. Once you've used a keyword argument, all subsequent arguments must also be keyword arguments.

    ```python
    def create_user(username, email, password, active=True):
        print(f"User: {username}, Email: {email}, Active: {active}")

    create_user("user123", "user@example.com", password="securepwd") # Valid
    create_user("user123", password="securepwd", email="user@example.com") # Invalid (positional arg 'email' after keyword arg 'password')
    ```

In summary, the most flexible way to pass arguments is using keyword arguments, where order doesn't matter for the named parameters. However, positional arguments are often more concise for required parameters, but you must remember their defined order.

## Understanding `*args` and `**kwargs`

These are powerful special syntaxes in Python that allow a function to accept a **variable number of arguments**.

### 1. `*args` (Arbitrary Positional Arguments)

*   **Purpose:** Allows a function to accept an arbitrary (any) number of **positional arguments**.
*   **Mechanism:** When you use `*args` as a parameter in a function definition, all excess positional arguments passed to the function are collected into a **tuple**.
*   **Naming Convention:** You can use any name after the `*` (e.g., `*items`, `*values`), but `*args` is the widely accepted and recommended convention.

### 2. `**kwargs` (Arbitrary Keyword Arguments)

*   **Purpose:** Allows a function to accept an arbitrary (any) number of **keyword arguments**.
*   **Mechanism:** When you use `**kwargs` as a parameter, all excess keyword arguments passed to the function are collected into a **dictionary**.
*   **Naming Convention:** Similarly, you can use any name after the `**` (e.g., `**details`, `**options`), but `**kwargs` is the standard convention.

### Combining `*args` and `**kwargs`

You can use both `*args` and `**kwargs` in the same function signature. When doing so, they must follow a strict order:

1.  **Standard Positional Arguments** (e.g., `param1`, `param2`)
2.  **`*args`**
3.  **`**kwargs`**

**Example:**

```python
def my_function(fixed_arg, *args, **kwargs):
    print("Fixed Argument:", fixed_arg)
    print("Positional Arguments (*args):"), args)
    print("Keyword Arguments (**kwargs):"), kwargs)

my_function("First Value", "arg1", "arg2", item='apple', count=3, active=True)
```

In this example:

*   `fixed_arg` receives `"First Value"`.
*   `*args` collects `"arg1"` and `"arg2"` into a tuple `('arg1', 'arg2')`.
*   `**kwargs` collects `item='apple'`, `count=3`, and `active=True` into a dictionary `{'item': 'apple', 'count': 3, 'active': True}`.

This flexibility makes functions adaptable to various inputs without needing to define numerous parameters upfront.

````python
def process_order(order_id, *items, **details):
    print(f"Processing Order ID: {order_id}")

    if items:
        print("Items ordered:")
        for item in items:
            print(f"- {item}")

    if details:
        print("Additional Details:")
        for key, value in details.items():
            print(f"- {key}: {value}")

print("--- Example 1: Basic order ---")
process_order("A123", "Laptop", "Mouse")

print("\n--- Example 2: Order with details ---")
process_order("B456", "Keyboard", "Monitor", customer_name="Alice", shipping_address="123 Main St")

print("\n--- Example 3: Only fixed and keyword args ---")
process_order("C789", payment_method="Credit Card", express_shipping=True)

print("\n--- Example 4: Only fixed arg ---")
process_order("D001")
````

## Unpacking Arguments with `*` and `**`

Beyond defining functions that accept arbitrary arguments (which we discussed with `*args` and `**kwargs`), the `*` and `**` operators have another powerful use: **unpacking iterables and dictionaries into function arguments** when calling a function.

### 1. Unpacking Lists or Tuples with `*` (for Positional Arguments)

*   **Purpose:** If you have a list or tuple of values that correspond to the positional parameters of a function, you can use the `*` operator to automatically unpack them.
*   **Mechanism:** When `*iterable` is used in a function call, Python treats each item in the `iterable` as a separate positional argument.
*   **Benefit:** Avoids manually passing each item as a distinct argument.

### 2. Unpacking Dictionaries with `**` (for Keyword Arguments)

*   **Purpose:** If you have a dictionary where keys match parameter names and values are the argument values, you can use `**` to unpack them.
*   **Mechanism:** When `**dictionary` is used in a function call, Python treats each key-value pair in the `dictionary` as a separate keyword argument.
*   **Benefit:** Ideal for functions that take many keyword arguments, especially when those arguments are dynamically generated or stored in a dictionary.

Let's look at your examples to see this in action.

````python
# --- Unpacking Lists/Tuples with * ---
print("\n--- Unpacking Lists with * ---")
def my_sum_function(a, b, c):
  print(f"Received: a={a}, b={b}, c={c}")
  return a + b + c

numbers = [1, 2, 3]
print(f"List to unpack: {numbers}")
result = my_sum_function(*numbers) # Same as: my_sum_function(1, 2, 3)
print(f"Result of my_sum_function(*numbers): {result}")

# --- Unpacking Dictionaries with ** ---
print("\n--- Unpacking Dictionaries with ** ---")
def my_greet_function(fname, lname):
  print(f"Hello {fname} {lname}")

person = {"fname": "Emil", "lname": "Refsnes"}
print(f"Dictionary to unpack: {person}")
my_greet_function(**person) # Same as: my_greet_function(fname="Emil", lname="Refsnes")

# Example with a mix of regular and unpacked arguments
def user_info(id, **data):
    print(f"User ID: {id}")
    for key, value in data.items():
        print(f"  {key.capitalize()}: {value}")

user_profile = {'name': 'Alice', 'email': 'alice@example.com', 'age': 30}
print("\n--- User Info with mixed args ---")
user_info(101, **user_profile)
````

## Understanding Variable Scope: Local vs. Global

In Python, the **scope** of a variable determines where in your code that variable can be accessed or modified. When you define variables, they exist in either a **global scope** or a **local scope**.

### 1. Global Scope

*   **Definition:** Variables defined outside of any function.
*   **Accessibility:** Accessible from anywhere in the program, both inside and outside functions.
*   **Lifetime:** They exist as long as the program runs.

### 2. Local Scope

*   **Definition:** Variables defined inside a function.
*   **Accessibility:** Accessible only from within the function where they are defined.
*   **Lifetime:** They are created when the function is called and destroyed when the function finishes execution.

### The "Golden Rule" for Same Variable Names

As your W3Schools example beautifully illustrates:

*   If you have a variable `x` in the global scope and you define another variable `x` *inside* a function, Python treats them as **two entirely separate variables**.
*   The `x` inside the function is a **local variable**, and any modifications to it will not affect the global `x`.
*   Inside the function, the local `x` takes precedence. If you reference `x` within that function, Python will look for the local `x` first.

### Using the `global` Keyword (to modify global variables)

If you explicitly want to modify a global variable from within a function, you must use the `global` keyword:

```python
x = 10 # Global x

def my_function():
    global x # Declare intent to modify the global x
    x = 20 # This now changes the global x, not a new local one

my_function()
print(x) # Output: 20
```

Let's run your example and then a demonstration of `global`.

````python
# Your example from W3Schools:
x = 300 # Global x

def myfunc():
  x = 200  # Local x (a new variable, does not affect global x)
  print(f"Inside myfunc, local x: {x}")

myfunc()
print(f"Outside myfunc, global x: {x}")

# --- Demonstration of the 'global' keyword ---
print("\n--- Using 'global' keyword ---")
y = 100 # Global y

def change_global_y():
    global y # Declare that we intend to modify the global 'y'
    y = 50   # This now modifies the global 'y'
    print(f"Inside change_global_y, y: {y}")

print(f"Global y before function call: {y}")
change_global_y()
print(f"Global y after function call: {y}")

# --- Attempting to read a global variable without modifying it ---
print("\n--- Reading global variable ---")
z = 1000 # Global z

def read_global_z():
    print(f"Inside read_global_z, z: {z}") # Reads global z, no 'global' keyword needed for reading

read_global_z()
print(f"Outside read_global_z, z: {z}")
````

## Nested Functions and the `nonlocal` Keyword

Sometimes you'll have functions defined inside other functions, known as **nested functions** or **inner functions**. In such cases, a variable might exist in an outer function's scope but not be global.

### The `nonlocal` Keyword

*   **Purpose:** The `nonlocal` keyword is used inside a nested function to indicate that a variable refers to a variable in the nearest **enclosing (outer) scope** that is *not* the global scope.
*   **Behavior:** It allows you to modify a variable from an outer, non-global scope within an inner function.
*   **Contrast with `global`:** `global` is for modifying variables in the module's global scope, while `nonlocal` is for modifying variables in an intermediate (enclosing) function's scope.

**Your `nonlocal` example:**

```python
def myfunc1():
  x = "Jane" # 'x' here is in the enclosing scope for myfunc2
  def myfunc2():
    nonlocal x # Declares intent to modify 'x' from myfunc1's scope
    x = "hello"
  myfunc2()
  return x

print(myfunc1()) # Output: hello
```

Without `nonlocal x` in `myfunc2()`, assigning `x = "hello"` would create a *new local* `x` inside `myfunc2`, leaving `myfunc1`'s `x` as "Jane".

### The LEGB Rule: How Python Finds Variables

Python uses a specific order to search for variable names (resolve scopes). This is known as the **LEGB Rule**:

1.  **L**ocal: The current function scope. (e.g., `x` inside `myfunc2` without `nonlocal`).
2.  **E**nclosing: Any enclosing (outer) function scopes, from innermost to outermost. (e.g., `x` in `myfunc1` when accessed by `myfunc2` with `nonlocal`).
3.  **G**lobal: The top-level module scope. (e.g., `x` outside all functions).
4.  **B**uilt-in: Python's pre-defined names (like `print`, `len`, `int`).

When a variable is accessed, Python checks these scopes in this exact order. The first match it finds is used.

**Your LEGB example:**

```python
x = "global"

def outer():
  x = "enclosing"
  def inner():
    x = "local"
    print("Inner:", x) # Accesses local 'x'
  inner()
  print("Outer:", x) # Accesses enclosing 'x'

outer()
print("Global:", x) # Accesses global 'x'
```

Let's run these examples to observe how `nonlocal` works and how the LEGB rule determines which `x` is printed at each stage.

````python
# --- nonlocal Keyword Example ---
print("\n--- nonlocal Keyword Demo ---")
def myfunc1():
  x = "Jane"  # Variable in the enclosing scope
  def myfunc2():
    nonlocal x # Declares intent to modify 'x' from myfunc1's scope
    x = "hello"
    print(f"  Inside myfunc2, x is now: {x}")
  print(f"Inside myfunc1 before myfunc2 call, x: {x}")
  myfunc2()
  print(f"Inside myfunc1 after myfunc2 call, x: {x}")
  return x

print(f"Result of myfunc1(): {myfunc1()}")

# --- LEGB Rule Example ---
print("\n--- LEGB Rule Demo ---")
x_legb = "global variable"

def outer_function():
  x_legb = "enclosing variable"
  def inner_function():
    x_legb = "local variable"
    print(f"  From inner_function (Local scope): {x_legb}")
  inner_function()
  print(f"From outer_function (Enclosing scope): {x_legb}")

outer_function()
print(f"From global scope: {x_legb}")

# Example of trying to assign to an outer scope without nonlocal
print("\n--- No nonlocal keyword ---")
def outer_no_nonlocal():
    val = "outer"
    def inner_no_nonlocal():
        val = "inner" # Creates a new local 'val', doesn't modify outer's 'val'
        print(f"  Inner: {val}")
    inner_no_nonlocal()
    print(f"Outer: {val}")

outer_no_nonlocal()
````

## Understanding Python Decorators

To grasp decorators, we first need to understand a core Python concept: **Functions are First-Class Objects.**

### 1. First Understand: Functions are Objects in Python

In Python, functions are not just blocks of code; they are treated like any other object. This means you can:

*   **Store them in variables:** Assign a function to a variable.
*   **Pass them to another function:** Use a function as an argument to another function.
*   **Return them from a function:** Have a function produce another function as its result.

### Example 1: Store function in a variable

````python
def greet():
    print("Hello")

x = greet # 'x' now refers to the same function object as 'greet'

x() # Calling x() is the same as calling greet()
````

### 2. Passing a Function as an Argument

You can pass function objects to other functions. This is a fundamental concept for decorators.

```python
def greet():
    print("Hello")

def call_function(func):
    func()

call_function(greet)
```

**Notice:** We pass `greet` (the function object itself), not `greet()` (which would execute the function and pass its return value).


````python
def greet():
    print("Hello")

def call_function(func):
    print("Inside call_function before executing passed function...")
    func() # Execute the passed function
    print("Inside call_function after executing passed function...")

call_function(greet)
````

### 3. Returning a Function from Another Function

Functions can also generate and return other functions. This is another key piece of the decorator puzzle.

```python
def outer():
    def inner():
        print("Inside inner")
    return inner

new_function = outer() # outer() executes and returns the 'inner' function object
new_function() # Now call the returned function
```

````python
def outer():
    print("Outer function is running...")
    def inner():
        print("Inner function is running...")
    return inner

new_function = outer() # outer() runs, prints its message, and returns 'inner'
print("Returned function is stored in new_function.")
new_function() # Now, calling new_function executes the 'inner' function
````

### 4. Now We Can Understand Decorators

A **decorator** is simply a function that:

*   **Takes another function as input.**
*   **Adds some extra functionality** to it (without modifying the original function's code directly).
*   **Returns a new function** (usually a wrapper function that includes the added functionality).

**General structure of a decorator:**

```python
def decorator(original_function):

    def wrapper(*args, **kwargs): # A wrapper function that will replace the original
        # --- Extra work before original_function ---
        result = original_function(*args, **kwargs) # Call the original function
        # --- Extra work after original_function ---
        return result # Return the result of the original function

    return wrapper # The decorator returns the new wrapper function
```

### 5. First Decorator Example

Let's say we have a simple function `say_hello()`:

```python
def say_hello():
    print("Hello")
```

We want to print something "Before function" and "After function" every time `say_hello` is called, *without* changing the `say_hello` function itself. We can achieve this using a decorator:

```python
def my_decorator(func):

    def wrapper():
        print("Before function")
        func() # Call the original function
        print("After function")
    return wrapper

def say_hello():
    print("Hello")

say_hello = my_decorator(say_hello) # Decorate the function manually
say_hello() # Call the *decorated* version
```

````python
def my_decorator(func):

    def wrapper():
        print("Before function")
        func()
        print("After function")
    return wrapper

def say_hello():
    print("Hello")

print("--- Manual Decoration ---")
say_hello = my_decorator(say_hello) # Here, say_hello is reassigned to the 'wrapper' function
say_hello()
````

### 6. The `@` Decorator Syntax (Syntactic Sugar)

The line `say_hello = my_decorator(say_hello)` is so common that Python provides a special syntax called the `@` (at) symbol to make it cleaner.

```python
@my_decorator
def say_hello():
    print("Hello")

say_hello()
```

This `@my_decorator` placed directly above the function definition is equivalent to `say_hello = my_decorator(say_hello)`.


````python
def my_decorator(func):

    def wrapper():
        print("Before function")
        func()
        print("After function")
    return wrapper

print("--- @ Decorator Syntax ---")
@my_decorator # This line automatically does: say_hello = my_decorator(say_hello)
def say_hello():
    print("Hello")

say_hello()
````

### 7. Decorator with Arguments (`*args` and `**kwargs`)

The previous decorator examples only work for functions that don't take any arguments. If the decorated function needs arguments, our `wrapper` function also needs to accept them.

This is where `*args` and `**kwargs` become essential in the `wrapper` function:

```python
def my_decorator(func):

    # The wrapper must accept any arguments that 'func' might take
    def wrapper(*args, **kwargs):
        print("Before function execution")
        result = func(*args, **kwargs) # Pass arguments to the original function
        print("After function execution")
        return result # Return the result of the original function

    return wrapper


@my_decorator
def add(a, b):
    print(f"Executing add({a}, {b})")
    return a + b


print("--- Decorator with Arguments ---")
sum_result = add(5, 10)
print(f"Sum Result: {sum_result}")

sum_result_kw = add(a=20, b=30)
print(f"Sum Result (keywords): {sum_result_kw}")
```

**Why `*args` and `**kwargs` in the wrapper?**

They allow the `wrapper` function to be a flexible intermediary, capable of accepting any number of:

*   **Positional arguments** (e.g., `add(5, 10)`)
*   **Keyword arguments** (e.g., `add(a=5, b=10)`)

This makes your decorators reusable for a wide variety of functions.

````python
def my_decorator(func):

    def wrapper(*args, **kwargs):
        print("Before function execution")
        result = func(*args, **kwargs)
        print("After function execution")
        return result

    return wrapper


@my_decorator
def add(a, b):
    print(f"Executing add({a}, {b})")
    return a + b

@my_decorator
def greet_person(name, greeting="Hello"):
    print(f"Executing greet_person({name}, {greeting})")
    return f"{greeting}, {name}!"


print("--- Decorator for `add` function ---")
sum_result = add(5, 10)
print(f"Sum Result: {sum_result}")

print("\n--- Decorator for `greet_person` function ---")
greet_result_pos = greet_person("Alice")
print(f"Greeting Result (positional): {greet_result_pos}")

greet_result_kw = greet_person(name="Bob", greeting="Hi")
print(f"Greeting Result (keyword): {greet_result_kw}")
````

````python
def changecase(func):
  def myinner():
    return func().upper()
  return myinner

@changecase
def myfunction():
  return "Hello Sally"

print(myfunction())
````

## Python Modules (from a W3Schools Perspective)

As W3Schools explains, a **module** in Python is simply a file containing Python code. It's a way to organize your code and reuse it across different Python scripts.

### What can a Module Contain?

A module can define functions, classes, and variables. The name of the module is the name of the file.

```python
# mymodule.py
```

### How to Use a Module

````python
%%writefile mymodule.py
# mymodule.py

def greeting(name):
  """This function returns a greeting message."""
  return f"Hello, {name}"

person1 = {
  "name": "John",
  "age": 36,
  "country": "Norway"
}

print("mymodule.py loaded successfully!") # This will print when the module is imported
````

Now, we can import `mymodule` and use its contents. Notice how the `print` statement inside `mymodule.py` executes when it's imported for the first time.

````python
import mymodule

# Accessing a function from the module
message = mymodule.greeting("Alice")
print(message)

# Accessing a dictionary from the module
print(f"Person's name: {mymodule.person1['name']}")
print(f"Person's age: {mymodule.person1['age']}")
````

### Different Ways to Import (Recap from earlier, aligned with W3Schools)

*   **`import module_name`**: Imports the entire module. Access with `module_name.item`.
*   **`from module_name import item_name`**: Imports specific items directly. Access with `item_name`.
*   **`import module_name as alias`**: Imports the module with an alias. Access with `alias.item`.
*   **`from module_name import *`**: Imports all items directly. Generally discouraged.

## Python Libraries: Theory and Examples

While "module" refers to a single `.py` file, the term "**library**" (or "package") often refers to a collection of related modules bundled together to provide a set of functionalities. A library can be a single module or a collection of modules and sub-packages arranged in a directory structure.

### What are Python Libraries?

Python libraries are collections of pre-written code that provide specific functionalities, saving developers from writing code from scratch for common tasks. They extend Python's core capabilities.

### Why use Libraries?

1.  **Reusability:** Leverage existing, tested, and optimized code.
2.  **Efficiency:** Speed up development by using ready-made solutions.
3.  **Community Support:** Many popular libraries have large communities, extensive documentation, and active maintenance.
4.  **Specialization:** Provides tools for specific domains (e.g., data science, web development, machine learning).

### Modules vs. Packages vs. Libraries

*   **Module:** A single Python file (`.py`).
*   **Package:** A directory containing multiple modules and an `__init__.py` file (which can be empty). Packages help in structuring the module namespace using dot notation (e.g., `package.module.function`).
*   **Library:** A broader term that refers to a collection of related modules and packages, often distributed together. A library provides a specific set of functionalities for a domain (e.g., NumPy for numerical operations, Pandas for data analysis).

### Examples of Popular Python Libraries

Let's look at some commonly used libraries and their basic usage.

#### 1. NumPy (Numerical Python)

A fundamental library for numerical computing, especially with arrays and matrices. It's much faster than standard Python lists for numerical operations.

````python
import numpy as np

# Create a NumPy array
my_array = np.array([1, 2, 3, 4, 5])
print(f"NumPy Array: {my_array}")

# Perform element-wise operations (much faster than lists for large datasets)
print(f"Array * 2: {my_array * 2}")

# Create a 2D array (matrix)
matrix = np.array([[1, 2], [3, 4]])
print(f"\nMatrix:\n{matrix}")
print(f"Sum of all elements in matrix: {np.sum(matrix)}")
````

#### 2. Pandas (Data Analysis Library)

Built on top of NumPy, Pandas provides data structures like DataFrames, making data manipulation and analysis easy and efficient.

````python
import pandas as pd

# Create a DataFrame (like a spreadsheet or SQL table)
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['New York', 'London', 'Paris']
}
df = pd.DataFrame(data)

print("DataFrame:")
print(df)

# Access a column
print(f"\nNames: {df['Name'].tolist()}")

# Filter data
print("\nPeople older than 28:")
print(df[df['Age'] > 28])
````

#### 3. Matplotlib (Plotting Library)

A comprehensive library for creating static, animated, and interactive visualizations in Python.

````python
import matplotlib.pyplot as plt

# Sample data
x = [1, 2, 3, 4, 5]
y = [2, 4, 5, 4, 6]

# Create a simple line plot
plt.plot(x, y)
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Simple Line Plot')
plt.grid(True)
plt.show()
````

These examples illustrate how libraries abstract complex functionalities into simple, reusable functions and objects, making Python incredibly powerful for a wide range of applications.
````python
class Student:

    # Class variable
    school = "ABC School"

    # Constructor
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    # Instance method
    def display(self):
        print(self.name, self.marks)

    # Class method
    @classmethod
    def change_school(cls, new_name):
        cls.school = new_name

    # Static method
    @staticmethod
    def is_pass(marks):
        return marks >= 40

    # String representation
    def __str__(self):
        return f"{self.name} - {self.marks}"


# Creating objects
s1 = Student("Ram", 90)
s2 = Student("Shyam", 80)

# Accessing attributes
print(s1.name)

# Calling methods
s1.display()

# Class variable
print(Student.school)

# Class method
Student.change_school("XYZ")

# Static method
print(Student.is_pass(50))

# Printing object
print(s1)
````
