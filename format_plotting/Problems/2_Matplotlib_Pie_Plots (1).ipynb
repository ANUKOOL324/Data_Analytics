```python
# Importing Libraries
import ast
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
df['job_skills'] = df['job_skills'].apply(lambda x: ast.literal_eval(x) if pd.notna(x) else x)
```

## 🟩 Health Insurance Pie Chart (2.14.1)

```python
df
df['job_health_insurance'].value_counts().plot(kind='pie',startangle =90 ,labels=['true','false'] , autopct = '%1.1f%%')
plt.title('Health Insurance')
plt.show()
```

## 🟨 Job Schedule Distribution (2.14.2)

```python
df
```

```python
import matplotlib.pyplot as plt

job_schedule_counts = (
    df['job_schedule_type']
    .value_counts()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(10, 8))

wedges, texts, autotexts = plt.pie(
    job_schedule_counts,
    labels=None,                 # remove direct labels from pie
    autopct='%1.1f%%',
    startangle=90,
    pctdistance=0.75,
    wedgeprops={'edgecolor': 'white'}
)

plt.title('Top 3 Job Schedule Type Distribution', fontsize=14)
plt.ylabel('')
plt.axis('equal')

plt.legend(
    wedges,
    job_schedule_counts.index,
    title="Job Schedule Type",
    loc="center left",
    bbox_to_anchor=(1, 0.5)
)

plt.show()
```
