```python
# Importing Libraries
import ast
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
df['job_skills'] = df['job_skills'].apply(lambda x: ast.literal_eval(x) if pd.notna(x) else x)
```

## 🟩 Data Scientist Hourly Salaries (2.17.1)

```python
df
```

```python
df['job_title_short'].unique()
```

```python
dh_salary = df[df['job_title_short']=='Data Scientist']['salary_hour_avg'].dropna()
dh_salary.reset_index(drop=True)
plt.hist(dh_salary, bins=10 , edgecolor='black')
plt.title('Data Scientist Hourly Salaries')
plt.xlabel('Salary')
plt.ylabel('Frequency')
plt.show()
```

---

## 🟨 Job Postings by Number of Skills (2.17.2)

```python
df
```

```python
# Count number of skills per job posting
df['num_skills'] = df['job_skills'].apply(lambda x: len(x) if isinstance(x, list) else 0)

# Plot histogram of number of skills per posting
plt.hist(df['num_skills'], bins=10, edgecolor='black')
plt.title('Job Postings by Number of Skills')
plt.xlabel('Number of Skills')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.hist(df[df['job_title_short']=='Data Engineer']['salary_hour_avg'].dropna(), bins=10, edgecolor='black')
plt.title('Data Engineer Hourly Salaries')
plt.xlabel('Salary')
plt.ylabel('Frequency')
plt.show()
```
