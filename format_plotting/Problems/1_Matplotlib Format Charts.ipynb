```python
# Importing Libraries
import ast
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
df['job_skills'] = df['job_skills'].apply(lambda x: ast.literal_eval(x) if pd.notna(x) else x)
```

## 🟩 Subplots with Bar Charts (2.13.1)

```python
df
```

```python
fig , ax = plt.subplots(1,2)

df['job_title_short'].value_counts().plot(kind='bar' , ax = ax[0])
df['job_schedule_type'].value_counts().head(3).plot(kind='bar' , ax = ax[1])

plt.suptitle('Job Titles and Schedules')
fig.tight_layout()
plt.show()
```

---

## 🟨 Subplots with Job Titles by Schedule (2.13.2)

```python
schedules = ['Full-time', 'Part-time']

#it is asking for the differnt job titles on different types of schedule
fig , ax = plt.subplots(1,2)
ax_flat = ax.flatten()

for id, st in enumerate(schedules):
  df_schedule = df[df['job_schedule_type']==st]
  df_schedule['job_title_short'].value_counts().plot(kind='bar', ax=ax_flat[id])
  ax_flat[id].set_title(f"JOB TYPE - {st}")
  ax_flat[id].set_xlabel('Job Title')
  ax_flat[id].set_ylabel('Count')

plt.suptitle('Job Titles by Schedule')

fig.tight_layout()
plt.show()
```

---

## 🟥 Top Skills for Job Roles Subplots (2.13.3)

```python
# Code Needed
job_roles = ['Senior Data Analyst', 'Senior Data Scientist', 'Senior Data Engineer']

df_explode = df.explode('job_skills')

fig,ax = plt.subplots(3,1 , figsize=(15 ,10))
for i , jr in enumerate(job_roles):
  df_jr = df_explode[df_explode['job_title_short']==jr]
  df_jr['job_skills'].value_counts().head(4).plot(kind='bar' , ax=ax[i])
  ax[i].set_title(jr)
  ax[i].set_xlabel('Skills')
  ax[i].set_ylabel('Count')

plt.suptitle("Top skills for different roles")
fig.tight_layout()
```
