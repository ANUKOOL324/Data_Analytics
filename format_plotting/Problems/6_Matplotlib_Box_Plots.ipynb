```python
# Importing Libraries
import ast
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
df['job_skills'] = df['job_skills'].apply(lambda x: ast.literal_eval(x) if pd.notna(x) else x)
```

## 🟩 Data Scientist Hourly Salary Box Plot (2.18.1)

```python
df
```

```python
import seaborn as sns
```

```python
sns.boxplot(data=df[df['job_title_short'] == 'Data Scientist'], x='salary_hour_avg')
```

---

## 🟨 Comparison of Hourly Salaries for Data Analyst (2.18.2)

```python
# Code Needed
countries = ['United States', 'Canada']
```

```python
fig , ax = plt.subplots(2,1,figsize=(15,15))
for i , c in enumerate(countries):
  sns.boxplot(df[(df['job_country']==c) & (df['job_title_short']=='Data Analyst')] ,ax=ax[i], x='salary_hour_avg')
  ax[i].set_title(c)

plt.show()
```

---

## 🟥 Comparison of Hourly Salaries in Canada (2.18.3)

```python
# Code Needed
job_titles = ['Data Analyst', 'Data Engineer', 'Data Scientist']
```

```python
fig , ax = plt.subplots(3,1,figsize=(15,15))

for i , j in enumerate(job_titles):
  sns.boxplot(df[(df['job_country']=='Canada') & (df['job_title_short']==j)] ,ax=ax[i], x='salary_hour_avg')
  ax[i].set_title(j)
```
