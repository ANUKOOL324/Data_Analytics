```python
# Importing Libraries
import ast
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
df['job_skills'] = df['job_skills'].apply(lambda x: ast.literal_eval(x) if pd.notna(x) else x)
```

## 🟩 Job Postings Trend for Cloud Engineers (2.16.1)

```python
cloud_engineer_df = df[df['job_title_short'] == 'Cloud Engineer'].copy()

# Aggregate job postings by date
job_postings_trend = cloud_engineer_df['job_posted_date'].dt.month.value_counts().sort_index()
job_postings_trend

# Plot the trend
plt.figure(figsize=(12, 6))
plt.plot(job_postings_trend.index, job_postings_trend.values , marker='o' ,linewidth=2)

plt.title('Job Postings Trend for Cloud Engineers')
plt.xlabel('month')
plt.ylabel('Number of Job Postings')
plt.grid(True)
plt.tight_layout()
plt.show()
```
