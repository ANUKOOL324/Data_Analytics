```python
!pip install -U datasets
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

## 🟩 Bar Chart with Labels (1.30.1)

```python
df
```

```python
job_country = df['job_country'].value_counts().head(5).sort_values()
```

```python
job_country
```

```python
import matplotlib.pyplot as plt
```

```python
plt.bar(job_country.index, job_country)
plt.title("jobs per country")
plt.xlabel("country")
plt.ylabel("No of Jobs")
plt.show()
```

---

## 🟨 Horizontal Bar Chart with Labels (1.30.2)

```python
plt.barh(job_country.index, job_country)
plt.title("jobs per country")
plt.ylabel("country")
plt.xlabel("No of Jobs")
plt.show()
```

---

## 🟥 Line Chart with Labels (1.30.3)

```python
plt.plot(job_country.index, job_country)
plt.title("jobs per country")
plt.xlabel("country")
plt.ylabel("No of Jobs")
plt.show()
```
