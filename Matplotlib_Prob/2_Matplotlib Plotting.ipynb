```python
!pip install -U datasets
import pandas as pd
from datasets import load_dataset

dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
df
```

## 🟩 Job Country Bar Chart (1.29.1)

```python
import matplotlib.pyplot as plt
```

```python
job_country = df['job_country'].value_counts().sort_values(ascending=False).head(5)
```

```python
job_country
```

```python
plt.bar(job_country.index,job_country)
plt.xticks(rotation=45 , ha='right')
plt.show()
```

## 🟨 Average Salary Horizontal Bar Chart (1.29.2)

```python
df
```

```python
avg_sal = df.groupby('job_title_short')['salary_year_avg'].mean().sort_values()
avg_sal
```

```python
#avg_sal.plot(kind='barh')
plt.barh(avg_sal.index,avg_sal)
plt.show()
```

---

## 🟥 Average Salary Trend Line Plot (1.29.3)

```python
plt.plot(avg_sal.index , avg_sal)
plt.xticks(rotation=45 ,ha='right')
plt.show()
```
