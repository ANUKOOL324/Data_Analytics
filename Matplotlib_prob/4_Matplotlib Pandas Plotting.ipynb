```python
!pip install -U datasets
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

## 🟩 Job Posting Location Bar Plot (1.31.1)

```python
df
```

```python
#job_location = df.groupby("job_location").size().sort_values(ascending=False).head(10)
job_location = df['job_location'].value_counts().head(10)
job_location
```

```python
plt.bar(job_location.index, job_location.values)
plt.xlabel('Location')
plt.xticks(rotation=45,ha='right')
plt.ylabel('Number of Jobs')
plt.title('Top 10 Job Locations')
plt.show()
```

## 🟨 Job Postings Trend Line Plot (1.31.3)

```python
df
```

```python
job_posting = df['job_posted_date'].dt.month.value_counts().sort_index()
job_posting
```

```python
plt.plot(job_posting.index,job_posting)
plt.xlabel('Month')
plt.ylabel('Number of Jobs')
plt.title('Job Postings Trend')
plt.show()
```
