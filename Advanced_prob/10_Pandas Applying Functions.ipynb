```python
# Importing Libraries
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

## 🟩 Convert Date to String (2.10.1)

```python
df['job_posted_date']
```

```python
def change_to_string(s):
    return s.strftime('%y-%m-%d')
```

```python
df['job_posted_date_str'] = df['job_posted_date'].apply(change_to_string)
df['job_posted_date_str']
df
```

```python
df.drop(
    columns='job_posted_date_str',
    inplace=True
)
```

```python
df
```

---

## 🟨 Days Since Posted (2.10.2)

```python
from datetime import date
```

```python
def days_since_posted(posted_date):
  return (date.today() - posted_date.date()).days
```

```python
df['days_since_posted']=df['job_posted_date'].apply(days_since_posted)
df['days_since_posted']
```

## 🟥 Salary Category (2.10.3)

```python
df
```

```python
def category_of_salary(salary):
  if salary < 20000:
    return 'Low'
  elif salary < 350000:
    return 'Medium'
  else:
    return 'High'
```

```python
df['category_of_salary']=df['salary_year_avg'].apply(category_of_salary)
```
