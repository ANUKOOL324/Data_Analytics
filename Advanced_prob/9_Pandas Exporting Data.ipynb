```python
# Importing Libraries
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

## 🟩 Export to CSV (2.9.1)

```python
df.to_csv('data_jobs.csv', index=False)
```

## 🟨 Filter and Export to CSV (2.9.2)

```python
filter_rows = df[df['job_title_short']=='Data Analyst']
filter_rows
```

```python
filter_rows.to_csv('filter_data_roles.csv' , index=False)
```
