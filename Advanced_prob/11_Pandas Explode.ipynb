```python
# Importing Libraries
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

## 🟨 Bar Plot of Top Skills for Cloud Engineers (2.11.1)

```python
df
```

```python
df_cloud = df[df['job_title_short']=='Cloud Engineer']
df_cloud
```

```python
import ast


def process_skill_entry(entry):
    # Case 1: If the entry is already a Python list, return it directly.
    if isinstance(entry, list):
        return entry

    # Case 2: If the entry is a string, attempt to parse it.
    if isinstance(entry, str):
        try:
            # Try to safely evaluate the string as a Python literal (e.g., "['skill1', 'skill2']").
            evaluated = ast.literal_eval(entry)
            # If successful, ensure the result is a list. If literal_eval returned a scalar (e.g., 'python' from "'python'"),
            # wrap that scalar in a list.
            return evaluated if isinstance(evaluated, list) else [evaluated]
        except (ValueError, SyntaxError):
            # If ast.literal_eval fails (meaning it's a simple string like 'python', not "['python']"),
            # treat the string itself as a single skill and wrap it in a list.
            return [entry]

    # Case 3: If the entry is a missing value (None or numpy.nan), return an empty list.
    if pd.isna(entry):
        return []

    # Fallback: For any other unexpected type, convert it to a string and wrap in a list.
    return [str(entry)]

# Apply the processing function to the 'job_skills' column
df_cloud['job_skills'] = df_cloud['job_skills'].apply(process_skill_entry)
```

```python
df_explode = df_cloud.explode('job_skills')
df_explode
```

```python
df_explode['job_skills'].value_counts().head().plot(kind='barh')
```
