```python
# Importing Libraries
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

## 🟩 Drop Missing Values (2.3.1)

```python
df = df.dropna();
```

---

## 🟩 Random Sample of DataFrame (2.3.2)

```python
df.sample(3)#random 3 samples
```

---

## 🟨 Fill Missing Salaries (2.3.3)

```python
df['salary_hour_avg']=df['salary_hour_avg'].fillna(df['salary_hour_avg'].median())
df['salary_year_avg']=df['salary_year_avg'].fillna(df['salary_year_avg'].median())
df
```
