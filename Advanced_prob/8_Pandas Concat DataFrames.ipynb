```python
# Importing Libraries
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

## 🟩 Concat Job Postings (2.8.1)

```python
job_jan = df[df['job_posted_date'].dt.month == 1]
job_feb = df[df['job_posted_date'].dt.month == 2]

concat_job_posting = pd.concat([job_jan , job_feb],ignore_index = False)
concat_job_posting
```

---

## 🟨 Concat Monthly Postings (2.8.2)

```python
df
```

```python
df['job_posted_month'] = df['job_posted_date'].dt.strftime('%B')
unique_months = df['job_posted_month'].unique()
unique_months
```

```python

dict_job_month = {month:df[df['job_posted_month']==month]for month in unique_months}
# dict_job_month['January']
concat_job_month = pd.concat(dict_job_month.values() , ignore_index=False)
concat_job_month
```

```python
df.drop('job_posted_month',inplace=True , axis=1)
df
```

## 🟨 Concat Job Types (2.8.3)

```python
jobs = df['job_title_short'].unique()
dict_jobs = {job:df[df['job_title_short']==job] for job in jobs}
# dict_jobs['Data Analyst']
```

```python
concat_jobs = pd.concat(dict_jobs.values() , ignore_index=False)
concat_jobs
```
