```python
# Importing Libraries
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

## 🟩 Merge US Jobs and Monthly Postings (2.7.1)

```python
df['job_posted_month'] = df['job_posted_date'].dt.month
df
us_jobs = df[df['job_country']=='United States']
us_jobs
monthly_postings = us_jobs.groupby('job_posted_month').size().reset_index(name='total_monthly_postings')#this will make this a dataframe
monthly_postings
pd.merge(us_jobs,monthly_postings , on='job_posted_month' , how='left')
```

---

## 🟨 Merge Non-Remote Jobs and Salary (2.7.2)

```python
non_remote = df[df['job_work_from_home']==False]

salary_summary = df.groupby('job_title_short', as_index=False)['salary_year_avg'].mean()


pd.merge(non_remote , salary_summary, on='job_title_short',how='left')
```

## 🟨 Merge Company Salary and Remote Count (2.7.3)

```python
remote_count = df.groupby('company_name' , as_index=False)['job_work_from_home'].count()
remote_count
company_salary = df.groupby('company_name', as_index=False)['salary_year_avg'].mean()
company_salary
pd.merge(remote_count , company_salary , on='company_name' , how='left')
```
