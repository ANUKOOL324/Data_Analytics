```python
# Importing Libraries
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

## 🟩 Set Index and Display Rows (2.5.1)

```python
df.set_index('job_title_short', inplace=True)
df.head()
```

---

## 🟩 Reset Index and Drop Existing (2.5.2)

```python
df.reset_index(drop=True , inplace=True)
df
```

---

## 🟩 Sort Index (2.5.3)

```python
# df_sort = df[df["job_country"]=='United States']
# df_sort

data = {
    "name": ["Ram", "Shyam", "Geeta"],
    "age": [22, 21, 20],
    "city": ["Delhi", "Mumbai", "Pune"]
}

dff = pd.DataFrame(data , ['3','2','1'])
dff
```

```python
dff.sort_index( inplace=True)
dff
```

## 🟨 Pivot Table Reset & Set Index (2.5.4)

```python
df_sal = df.pivot_table(values='salary_year_avg',index='job_title' , aggfunc='mean').head()
df_sal
#df_sal.index
```

```python
df_sal.reset_index(inplace=True)
df_sal
```

```python

df_sal.set_index('job_title' , inplace=True)
df_sal
```
