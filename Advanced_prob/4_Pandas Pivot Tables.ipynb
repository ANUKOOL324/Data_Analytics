```python
# Importing Libraries
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

## 🟩 Count Job Postings by Title and Country (2.4.1)

```python
df
```

```python
job_posting = df.pivot_table(values='job_posted_date',index='job_title_short',columns='job_country',aggfunc='count');
job_posting
```

---

## 🟨 Salary Stats by Job Title (2.4.2)

```python
salary_stats = df.groupby('job_title_short')['salary_year_avg'].describe();
display(salary_stats)
```

```python
salary_stat = df.pivot_table(values='salary_year_avg' , index='job_title_short',aggfunc=['min','max','count','median'])
salary_stat
```

---

## 🟥 Sum Salary by Title and Country (2.4.3)

```python
sum_salary = df.pivot_table(values='salary_year_avg',index='job_title_short',columns='job_country',aggfunc='sum' , fill_value=0)
sum_salary
```
