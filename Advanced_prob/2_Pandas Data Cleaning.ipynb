```python
# Importing Libraries
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

## 🟩 Remove Rows with Missing Values (2.2.1)

```python
bf = df.copy()
bf
```

```python
bf.dropna(inplace=True)
bf
```

## 🟨 Remove Duplicate Rows (2.2.2)

```python
df.duplicated()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates().reset_index(drop=True)
df
```

---

## 🟨 Fill Missing Values in Column (2.2.3)

```python
df
```

```python
df['salary_hour_avg']=df['salary_hour_avg'].fillna(df['salary_hour_avg'].median())#this col nan will be assigned the median of the whole column
df
```

```python
#if whole df nan to a value then
df.fillna(0)
```
